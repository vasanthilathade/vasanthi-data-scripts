```python
def my_sub_strings_palindrome(s):
    n = len(s)
    result = set()

    for i in range(n):
        for j in range(i + 1, n + 1):
            sub = s[i:j]

            is_palindrome = True
            for k in range(len(sub) // 2):
                if sub[k] != sub[-(k + 1)]:
                    is_palindrome = False
                    break

            if is_palindrome:
                result.add(sub)

    return result


s = "ababa"
my_sub_strings_palindrome(s)
```

```python
string="abcDfeGHIjk1234567HJKL"
# encode the string
# replace lower case as # and upper case * and digit as $
def my_hash(string):
    result=""
    for ch in string:
        if 'a' <= ch <= 'z':
            result += '#'
        elif 'A' <= ch <= 'Z':
            result += '*'
        elif '0' <= ch <= '9':
            result += '$'
    return result

my_hash(string)
```

```python
string = "abcDfeGHIjk1234567HJKL"

def my_hash(string):
    result = ""
    for ch in string:
        ascii_val = ord(ch)

        if 97 <= ascii_val <= 122:
            result += '#'
        elif 65 <= ascii_val <= 90:
            result += '*'
        elif 48 <= ascii_val <= 57:
            result += '$'

    return result

print(my_hash(string))
```

```python
# upper case # and lower case @ space with $ and num with  number % - line wise replacing

sss = """This is a python code
we are discussing about the string manipulation
this is 2nd session"""

output = ""
for line in sss.split("\n"):
    for ch in line:
        if ch.islower():
            output += "@"
        elif ch.isupper():
            output += "#"
        elif ch.isdigit():
            output += "%"
        elif ch == " ":
            output += "$"
    output+="\n"


print(output)

```

```python
from collections import defaultdict

def string_anagram(str1, str2):
    if len(str1) != len(str2):
        return False

    dict1 = defaultdict(int)
    dict2 = defaultdict(int)

    for ch in str1:
        dict1[ch] += 1

    for ch in str2:
        dict2[ch] += 1

    print(dict1)
    print(dict2)
    return dict1 == dict2


string_anagram('listen', 'silent')
```
