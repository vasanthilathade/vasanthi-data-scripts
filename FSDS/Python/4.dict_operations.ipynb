```python
#approach 2
def two_sum(lst, target):
    result = []
    visited={}
    for index, element in enumerate(lst):
        diff = target - element
        if diff in visited:
            result.append([index,visited[diff]])
        visited[element]=index
    return result


two_sum([1,2,3,4,5], 3)
```

```python
# return the common element between 2 list
lst1=[2,3,4,5,6]
lst2=[5,6,7,8,9]
def my_common_elements(lst1, lst2):
    dict2 = {}
    result = []
    for element in lst2:
        dict2[element] = 1

    for element in lst1:
        if element in dict2:
            result.append(element)

    return result

my_common_elements(lst1, lst2)
```

```python
def my_dict_popitem(dict1, value):
    new_dict={}
    for index, element in dict1.items():
        if element != value:
            new_dict[index] = element
    return new_dict

dict1={1: 0, 2: 11, 3: 22, 4: 13, 5: 42}
print(my_dict_popitem(dict1, 42))

```

```python
# remove the last element of the dict without using the pop item
def my_dict_popitem(dict1):
    last_index = -1
    for index in dict1:
        last_index = index
    if last_index:
        deleted_element=dict1[last_index]
        del dict1[last_index]
        return  dict1,deleted_element

dict1={1: 0, 2: 11, 3: 22, 4: 13, 5: 42}
print(my_dict_popitem(dict1))

```

```python
# merge 2 dictionaries
dict1={
    1:100,
    2:200,
    3:300
}
dict2={
    4:400,6:600,7:700
}
def my_dict_merge(dict1, dict2):
    result={}
    for index,element in dict1.items():
        result[index]=element
    for index,element in dict2.items():
        result[index]=element
    return result
my_dict_merge(dict1,dict2)
```

```python
# sort dict based on the values wihtout using lambda and iter functions
dict1 = {2: 300, 4: 500, 5: 50, 12: 47, 7: 11}


def my_dict_sort(dict1):
    result = {}
    temp = {}
    for key, value in dict1.items():
        temp[value] = key
    while dict1 != {}:
        min_element = min(dict1.values())
        result[temp[min_element]] = min_element
        del dict1[temp[min_element]]
    return result


print("input dict1: ", dict1)
print("sorted dict: ", my_dict_sort(dict1))


```

```python
# sort dict based on the values wihtout using lambda and iter functions
dict1 = {2: 300, 4: 500, 5: 50, 12: 47, 7: 11}


def my_dict_sort(dict1):
    result = {}

    while dict1:
        min_value = None
        index_value = None
        for key, value in dict1.items():
            if min_value is None or value < min_value:
                min_value = value
                index_value = key
        result[index_value] = min_value
        del dict1[index_value]
    return result


print("input dict1: ", dict1)
print("sorted dict: ", my_dict_sort(dict1))


```

```python
"""
Dictionary Comprehension in Python
----------------------------------

Dictionary comprehension is a concise way to create dictionaries.
It is similar to list comprehension, but it generates key–value pairs.

General Syntax:
{key: value for item in iterable if condition}
"""
```

```python
# Create a dictionary where keys are numbers and values are their squares
{i: i**2 for i in range(3)}
```

```python
# Include only even numbers as keys
{i: i**2 for i in range(10) if i % 2 == 0}
```

```python
# List of words
lst = ["This", "is", "a", "dict", "comprehension"]

# Dictionary with word as key and its length as value
{key: len(key) for key in lst}
```

```python
keys = ["name", "age", "city"]
values = ["Alice", 25, "Delhi"]

# Combine two lists into a dictionary
{keys[i]: values[i] for i in range(len(keys))}
```

```python
# Original dictionary
d = {"a": 1, "b": 2, "c": 3}

# Swap keys and values
{value: key for key, value in d.items()}
```

```python
# Filter items where value is greater than 2
d = {"a": 1, "b": 3, "c": 5, "d": 2}

{key: value for key, value in d.items() if value > 2}
```

```python
# Count occurrences of each character
s = "dictionary"

{char: s.count(char) for char in s}
```

```python
# Apply condition on values
scores = {"Math": 45, "Science": 80, "English": 60}

# Pass or Fail based on score
{subject: "Pass" if marks >= 50 else "Fail" for subject, marks in scores.items()}
```
