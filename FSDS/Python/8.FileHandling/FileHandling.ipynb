**What is File Handling?**

File handling is a mechanism that allows a user to perform operations on files—such as creating, opening, reading, writing, and closing files—using a program.

In file handling, data can be stored in two types of files:

1. **Text Files**
   These files store data in text format (human-readable characters).

2. **Binary Files**
   These files store data in binary format, such as audio, video, images, and other non-text data.

**File Functions:**

* **`open()`** → Used to open a file.
* **`close()`** → Used to close a file.

**What are File Modes?**
File modes specify the operation to be performed on a file.

* **`'r'`** → Opens a file in read mode. (File must exist)
* **`'w'`** → Opens a file in write mode. (Creates a new file or overwrites existing content)
* **`'a'`** → Opens a file in append mode. (Adds data at the end of the file)
* **`'x'`** → Opens a file in exclusive creation mode. (Fails if the file already exists)
* **`'r+'`** → Opens a file for both reading and writing.
* **`'w+'`** → Opens a file for reading and writing. (Overwrites existing content)
* **`'a+'`** → Opens a file for reading and appending.

**Binary Modes:**
'b'-> binary
* **`'rb'`** → Read binary file
* **`'wb'`** → Write binary file
* **`'ab'`** → Append binary file
