**What is File Handling?**

File handling is a process that allows users to perform operations on files—such as creating, opening, reading, writing, and closing files—through a program interface.

In file handling, data can be stored in two types of files:

1. **Text Files**
   These files store data in text format (readable characters).

2. **Binary Files**
   These files store data in binary format, such as audio, video, images, and other non-text data.
