```python
'''
It is a consized way of writing the code
Whatever the things we will able to do using the normal looping we do that with list comphrension
syntax:
[each_item for item in list]
'''
```

```python
# print first 10 numbers
s=[element for element in range(10)]
print(s)
```

```python
[i**2 for i in range(10)]
```

```python
[i for i in range(10) if i%2==0]
```

```python
lst = [1, 2, 3, 4, 5, 6, 7, 8, 9]

result = ['even' if element % 2 == 0 else 'odd' for element in lst]
print(result)
```

```python
lst_words=["This", "is", "a", "list", "comprehension"]
s=[word.upper() for word in lst_words]
print(s)
```

```python
for i in range(3):
    for j in range(3):
        print(i,j)
[(i, j) for i in range(3) for j in range(3)]
```

```python
for i in range(3):
    for j in range(3):
        print(i,j)
[(i, j)  for i in range(3) for j in range(3) if i%2==0 and j%2==0]
```

```python
[(i, j) for i in range(3) for j in range(3) if i % 2 == 0 and j % 2 != 0]
```
